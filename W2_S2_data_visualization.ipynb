## Task 1 — Load and inspect

**Copilot prompt idea:** `# Load week2.csv and show the first five rows`

Survey columns we use today (full wording on slides):

| Column | Survey question |
|--------|------------------|
| `support_info` | *To what extent do you support this policy?* (1–5: Strongly oppose → Strongly support) |
| `Distance_artificial` | Derived distance to nearest recycling facility (m) — not a single form item |
| `recycling_effort` | *How much time and effort has it typically taken you to recycle in the past?* (sorting, cleaning, delivery, etc.) — coded 1–4: Never → Significant effort |


```python
from pathlib import Path

DATA_FILE = "week2.csv"

def load_week2_data():
    for p in [Path(DATA_FILE), Path.cwd() / DATA_FILE]:
        if p.is_file():
            print(f"Using data file: {p.resolve()}")
            return pd.read_csv(p)
    raise FileNotFoundError(
        f"Could not find {DATA_FILE}. Open a Codespace from your fork — "
        "do not use a path from your laptop."
    )

df = load_week2_data()
print("shape:", df.shape)
df[["support_info", "Distance_artificial", "recycling_effort"]].head()
```

```python
# Copilot: display df.info() or describe() for the three columns above
df[["support_info", "Distance_artificial", "recycling_effort"]].describe()
```

## Task 2 — Categorical data: support level

**Goal:** One frequency table + one **bar chart** for `support_info`.

**Survey item:** *To what extent do you support this policy?*  
**Scale:** 1 Strongly oppose … 5 Strongly support

**Copilot prompts (examples):**
- `# value_counts for support_info`
- `# bar chart for support_info with labels 1=Strongly oppose ... 5=Strongly support`

**Discuss:** What is one thing this chart shows? What does it **not** show (e.g. causation, whole HK population)?


```python
# Frequency table
support_counts = df["support_info"].value_counts().sort_index()
print(support_counts)

# Bar chart
labels = {
    1: "1 Strongly oppose",
    2: "2 Oppose",
    3: "3 Neutral",
    4: "4 Support",
    5: "5 Strongly support",
}
order = sorted(df["support_info"].dropna().unique())
plot_labels = [labels.get(int(x), str(x)) for x in order]
counts = [support_counts.get(x, 0) for x in order]

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(plot_labels, counts, color="steelblue")
ax.set_title("Support for MSW charging (survey sample)")
ax.set_ylabel("Count")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```

## Task 3 — Continuous data: distance to recycling

**Goal:** Summary statistics + **histogram** (or box plot) for `Distance_artificial`.

**Copilot prompt:** `# histogram of Distance_artificial`

**Governance tie-in:** Check the slide — how was “distance” defined? Would a policymaker ask about missing locations?


```python
print(df["Distance_artificial"].describe())

fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(df["Distance_artificial"].dropna(), bins=15, color="seagreen", edgecolor="white")
ax.set_xlabel("Distance to nearest recycling facility (m)")
ax.set_ylabel("Count")
ax.set_title("Distribution of distance (survey sample)")
plt.tight_layout()
plt.show()
```

## Task 4 — Relationship: distance vs recycling effort

**Goal:** One **scatter plot** (jitter optional) for `Distance_artificial` vs `recycling_effort`.

**Recycling effort scale (1–4):** 1 Never recycled · 2 Slight effort · 3 Moderate effort · 4 Significant effort

**Copilot prompt:** `# scatter plot Distance_artificial vs recycling_effort`

**Pause:** Does this plot prove that living farther away *causes* less recycling? Write one line: "We would not claim ... because ...".


```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(
    df["Distance_artificial"],
    df["recycling_effort"] + (pd.Series(range(len(df))) % 5) * 0.05,  # light jitter on y
    alpha=0.7,
)
ax.set_xlabel("Distance_artificial (m)")
ax.set_ylabel("recycling_effort (1–4, jittered for visibility)")
ax.set_title("Distance vs recycling effort — association only")
plt.tight_layout()
plt.show()
```
